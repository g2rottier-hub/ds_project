```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
a = np.array([1,2,5,4,5])
print(a)
```

```python
plt.plot(a)
plt.show()
```

```python
x_values=[10,20,30,40,50]
y_values=[1,2,3,1,3]

plt.plot(x_values,y_values)
plt.show
```

```python
#days is consecutive numbers from 1 to 7 (x-axis)
# temperature for days (y-axis)
#label my x-axis and y-axis
#title my plot

days=np.arange(1,8)
temperature=np.array([36,40,60,54,70,55,44])
plt.plot(days,temperature)
plt.xlabel("days of the week")
plt.ylabel("Temperature")
plt.title("Weekly Temperature")
plt.show()
```

```python
# same plot but i want to color the line with a different color
days=np.arange(1,8)
temperature=np.array([36,40,60,54,70,55,44])

plt.plot(days,temperature,color="orange")
plt.xlabel("days of the week")
plt.ylabel("Temperature")
plt.title("Weekly Temperature")
plt.show()
```

```python
# try a different line style and line width

plt.plot(days,temperature,color="orange",linestyle="--",linewidth=2)
plt.xlabel("days of the week")
plt.ylabel("Temperature")
plt.title("Weekly Temperature")
plt.show()
```

```python
#multiple line plots in the same plot

days=np.arange(1,8)
london_temp=np.array([36,40,60,54,70,55,44])
new_york_temp=np.array([50,55,52,58,65,70,68])

plt.plot(days,london_temp,color="blue",linestyle="-",label="London")
plt.plot(days,new_york_temp,color="red",linestyle="--",label="New York")

plt.xlabel("Day of the Week")
plt.ylabel("Temeprature (F)")
plt.title("Weekly Temperature: LOndon vs New York")
plt.legend()
plt.show()
```

```python
#marker style

days=np.arange(1,8)
temperature=np.array([36,40,60,54,70,55,44])
plt.plot(days,temperature,marker="o",markersize=8)
plt.xlabel("days of the week")
plt.ylabel("Temperature")
plt.title("Weekly Temperature")
plt.show()
```

```python
#scatter plot

study_hours = np.arange(1,11)
test_scores=np.array([45,50,55,58,65,70,72,92,80,85])
plt.scatter(study_hours,test_scores,c='brown',alpha=.8)
plt.show()
```

```python
#bar chart

subjects=['math','english','science','history','art']
scores=[88,72,95,60,85]

plt.bar(subjects, scores)
plt.xlabel('subjects')
plt.ylabel('score')
plt.title('Student scores by Subject')
plt.show()
```

```python
#histograms

scores=np.random.normal(loc=0,scale=10,size=1000)

plt.hist(scores,bins=40)
plt.xlabel('score')
plt.ylabel('Number of students')
plt.title('Distribution of scores')
plt.show()
```
